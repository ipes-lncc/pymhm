# Mixed H(div) elasticity families

The BDM, BDM-plus and BDM-double-plus families follow Table 1 of the [2021 weak-symmetry MHM paper](https://doi.org/10.1051/m2an/2021013). Fine-edge normal degree $k$ is independent of interior degree $k+n$, where $n=0,1,2$. Displacement and independent rotation use $P_{k+n-1}$. Local rigid rotations require $k+n\geq2$.

```python
import json
from pathlib import Path
from IPython.display import Image, display

root = Path.cwd()
while not (root / "pyproject.toml").exists():
    root = root.parent
```

The stress is the full Cauchy tensor. The skeleton carries negative normal traction; symmetry is imposed through rotation moments. Compliance uses deviatoric and spherical parts, and the infinite first Lamé modulus is an explicit supported limit. Polynomial patches, physical moments and four independently assembled DOLFINx/UFL operators verify these conventions.

```python
data = root / "examples/results/elasticity-families"
for path in sorted(data.glob("bdm*.json")):
    record = json.loads(path.read_text())
    print(path.stem, record["rows"][-1], "incompressible:", record["sweep"][-1]["stress_l2"])
```

Five mesh levels and five finite/infinite bulk-modulus values use an original bounded-force polynomial family. Each record states the macrotrace degree; different local families need not share that degree. These errors quantify the declared spaces and do not constitute a proof for arbitrary heterogeneous coefficients.
