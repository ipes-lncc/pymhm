# Mixed H(div) elasticity families

The BDM, BDM-plus and BDM-double-plus families follow Table 1 of the [2021 weak-symmetry MHM paper](https://doi.org/10.1051/m2an/2021013). Fine-edge normal degree $k$ is independent of interior degree $k+n$, where $n=0,1,2$. Displacement and independent rotation use $P_{k+n-1}$. Local rigid rotations require $k+n\geq2$.

The PyMHM distribution contains only the library. The first cell explicitly
downloads a checksum-verified companion archive and prepares the declared data
using its local Python helpers. You can inspect these support files in `ROOT`.
Complete studies and historical replay retain their existing opt-in flags.
