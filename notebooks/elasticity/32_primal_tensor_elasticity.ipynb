## Errors and the face indicator

The analytical displacement is $u=(\sin(\pi x)\sin(\pi y),\sin(2\pi x)\sin(\pi y))$. Stiffness is constant or multiplied by $1+x+2y$; the load includes its derivatives. The L17 face indicator assumes a declared ellipticity bound and full Dirichlet boundaries. Its reliability theorem requires exact local lifts, so the finite-local indicator below does not certify the complete two-level error.
