# General-tensor tetrahedral elasticity
The primal MHM formulation retains three translations and three rotations. Continuous P2/P3 local displacement fields use P1 vector traces. Stress is not generally H(div)-conforming; the finite-Lamé study is not a proof of locking-free behavior.

The original solenoidal manufactured displacement is the curl of $(0,0,\sin^2(\pi x)\sin^2(\pi y)\sin^2(\pi z))$. The anisotropic stiffness varies as $1+x+2y+3z$. Full DOLFINx/UFL operator tests cover P1–P4.

The sections evaluate the complete local displacement polynomials and physical stress on disconnected fine-tetrahedron display grids. Coincident interface points are kept separate. Signed component panels complement the displacement and stress magnitudes; these display samples are not the volume-norm quadrature.
