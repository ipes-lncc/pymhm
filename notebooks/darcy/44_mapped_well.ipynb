# Mapped hexahedral RT1 around a well
This example uses the Dupuit–Thiem physical parameters of Duran et al. (2019), §7.4/Table 1, on an explicitly declared polygonal reservoir. The 32-cell starting mesh has eight sectors and four logarithmically graded radial bands. The paper does not determine consistent numerical radii, so historical geometry identity is not asserted.

RT1 fluxes use the contravariant Piola map; pressure is the scalar pullback Q1. Traces prescribe reference-face flux densities. The four MHM macro levels share 16,384 fine hexahedra, ending at the classical conforming mixed method. A separate five-grid classical study compares physical volume errors with the analytical solution.

The PyMHM distribution contains only the library. The first cell explicitly
downloads a checksum-verified companion archive and prepares the declared data
using its local Python helpers. You can inspect these support files in `ROOT`.

The analytical pressure is imposed on both polygonal lateral boundaries; top and bottom are impermeable. The rate parameter is not a separate Neumann condition. All coefficients and errors retain SI units. Exact/numerical field panels have common scales and preserve one-sided values. Colored profile ticks identify actual macro intersections.

Pressure-tested divergence moments include fine-cell conservation. On nonaffine hexahedra, div(RT1) is Q1/det(J), not the ordinary pressure pullback Q1. Sampled positive Jacobians do not prove global validity of arbitrary distorted geometry.
